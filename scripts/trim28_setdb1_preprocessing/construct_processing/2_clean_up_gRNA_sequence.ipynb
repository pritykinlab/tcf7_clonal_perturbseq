```python
import pandas as pd

# Define the input and output file paths
input_file = "/Genomics/pritykinlab/share/perturbseq/fromMarc/new_data/trim28_setdb1_data/analysis/Lane1/feature_reference.csv"
output_file = "/Genomics/pritykinlab/share/perturbseq/fromMarc/new_data/trim28_setdb1_data/reference_genome/guide_seq.csv"

# Read the CSV file
df = pd.read_csv(input_file)

# Filter out rows with "Hashtag" in the 'name' column
df_filtered = df[~df['name'].str.contains('Hashtag')]

# Select relevant columns and rename them
df_relevant = df_filtered[['target_gene_name', 'name', 'sequence']]
df_relevant.columns = ['gene', 'guide_name', 'sequence']

# Save the filtered and formatted data to a new CSV file
df_relevant.to_csv(output_file, index=False)

print(f"Filtered and formatted data has been saved to {output_file}")
```
